```python
import numpy as np
import pandas as pd
```

```python
dataset = pd.read_csv("PlayTennis.csv")
```

```python
data = dataset.to_numpy()
```

```python
data = data[:,1:]
```

```python
print(data)
```

```python
P_H = []
p_h_n = 0
p_h_y = 0
for i in range(data.shape[0]):
    if data[i][4] == "No":
        p_h_n += 1
    else:
        p_h_y += 1
P_H.append(p_h_n / data.shape[0])
P_H.append(p_h_y / data.shape[0])
print(P_H)
```

```python
P_X1 = []
p_X1_1 = 0
p_X1_2 = 0
p_X1_3 = 0
for i in range(data.shape[0]):
    if data[i][0] == "Sunny":
        p_X1_1 += 1
    elif data[i][0] == "Overcast":
        p_X1_2 += 1
    else:
        p_X1_3 += 1
P_X1.append(p_X1_1 / data.shape[0])
P_X1.append(p_X1_2 / data.shape[0])
P_X1.append(p_X1_3 / data.shape[0])
print(P_X1) 
```

```python
P_X2 = []
p_X2_1 = 0
p_X2_2 = 0
p_X2_3 = 0
for i in range(data.shape[0]):
    if data[i][1] == "Hot":
        p_X2_1 += 1
    elif data[i][1] == "Mild":
        p_X2_2 += 1
    else:
        p_X2_3 += 1
P_X2.append(p_X2_1 / data.shape[0])
P_X2.append(p_X2_2 / data.shape[0])
P_X2.append(p_X2_3 / data.shape[0])
print(P_X2) 
```

```python
P_X3 = []
p_X3_1 = 0
p_X3_2 = 0
for i in range(data.shape[0]):
    if data[i][2] == "High":
        p_X3_1 += 1
    else:
        p_X3_2 += 1
P_X3.append(p_X3_1 / data.shape[0])
P_X3.append(p_X3_2 / data.shape[0])
print(P_X3) 
```

```python
P_X4 = []
p_X4_1 = 0
p_X4_2 = 0
for i in range(data.shape[0]):
    if data[i][3] == "Weak":
        p_X4_1 += 1
    else:
        p_X4_2 += 1
P_X4.append(p_X4_1 / data.shape[0])
P_X4.append(p_X4_2 / data.shape[0])
print(P_X4) 
```
