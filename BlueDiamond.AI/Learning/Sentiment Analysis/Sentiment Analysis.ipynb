```python
import numpy as np
import pandas as pd
import nltk as nltk
from nltk.tokenize import word_tokenize
from nltk.probability import FreqDist
from nltk.corpus import stopwords
import string
```

```python
dataset = pd.read_csv("FinancialData.csv")
dataset = dataset.to_numpy()
```

```python
print(dataset)
```

```python
text = ""
for i in range(dataset.shape[0]):
    text += dataset[i][0] + " "
print(text)
```

```python
stop = set(stopwords.words('english') + list(string.punctuation))
text_tokens = [i for i in word_tokenize(text.lower()) if i not in stop]
print(text_tokens)
```

```python
freqDist = FreqDist(text_tokens)
for key, value in freqDist.items():
    print(key, ' : ', value)
```

```python
classes = ["positive", "negative", "neutral"]
```

```python
def findNumDocsForClass(data, class_):
    N_class = 0
    for i in range(data.shape[0]):
        if data[i][1] == class_:
            N_class += 1
    return N_class
```

```python
def findSumOccurences(bigDocDict):
    sum_c_w_c = 0
    for word in bigDocDict.keys():
        sum_c_w_c += bigDocDict[word] + 1
    return sum_c_w_c
```

```python
def combineDocs(data, class_):
    bigDoc = ""
    for i in range(data.shape[0]):
        if data[i][1] == class_:
            bigDoc += data[i][0] + " "
    return bigDoc
```

```python
def TrainNaiveBayes(data, stop, classes, V):
    N_doc = data.shape[0]
    LogPriori = []
    LogLikelihood = []
    for class_ in classes:
        N_class = findNumDocsForClass(data, class_)
        logPriori = np.log(N_class / N_doc)
        bigDoc = combineDocs(data, class_)
        bigDocTokens = [i for i in word_tokenize(bigDoc.lower()) if i not in stop]
        bigDocDict = FreqDist(bigDocTokens)
        logLikelih = {}
        for word in V.keys():
            if bigDocDict.get(word) is not None:
                c_w_c = bigDocDict[word]
            else:
                c_w_c = 0
            sum_c_w_c = findSumOccurences(bigDocDict)
            logLikelihood_w_c = np.log((c_w_c + 1) / sum_c_w_c)
            logLikelih[word] = logLikelihood_w_c
        logLike = logLikelih.copy()
        LogLikelihood.append(logLike)
        logLikelih.clear()
        LogPriori.append(logPriori)
    return LogPriori, LogLikelihood 
```

```python
LogPriori, LogLikelihood = TrainNaiveBayes(dataset, stop, classes, freqDist)
```

```python
print(LogPriori, LogLikelihood) #Negative because Log[0,1] gives negative answers
```

```python
def TestNaiveBayes(TestDoc, data, LogPriori, LogLikelihood, classes, V):
    sum_logProb = []
    for i in range(len(classes)):
        sum_logProb.append(0)
    for i in range(len(classes)):
        sum_logProb[i] = LogPriori[i]
        for j in range(len(TestDoc)):
            word = TestDoc[j]
            if word in V:
                if LogLikelihood[i].get(word) is not None:
                    sum_logProb[i] += LogLikelihood[i][word]
    return np.argmax(sum_logProb)
```

```python
TestDoc = "The GeoSolutions technology will leverage Benefon 's GPS solutions by providing Location Based Search Technology , a Communities Platform , location relevant multimedia content and a new and powerful commercial model ."
TestDoc = TestDoc.split()
TestDoc = [x.lower() for x in TestDoc]
TestDoc_ = "A financial decline and bad performance ."
TestDoc_ = TestDoc_.split()
TestDoc_ = [x.lower() for x in TestDoc_]
class_ = TestNaiveBayes(TestDoc, dataset, LogPriori, LogLikelihood, classes, text_tokens)
class__ = TestNaiveBayes(TestDoc_, dataset, LogPriori, LogLikelihood, classes, text_tokens)
```
