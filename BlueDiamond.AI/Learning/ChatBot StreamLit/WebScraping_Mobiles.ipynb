```python
import requests
from bs4 import BeautifulSoup
import unicodedata
import json
```

```python
urls = [
    'https://www.whatmobile.com.pk/Samsung_Galaxy-Z-Fold-5',
    'https://www.whatmobile.com.pk/Samsung_Galaxy-Z-Fold-4-12GB',
    'https://www.whatmobile.com.pk/Samsung_Galaxy-S23-Ultra',
    'https://www.whatmobile.com.pk/Samsung_Galaxy-Z-Fold-4',
    'https://www.whatmobile.com.pk/Apple_iPhone-14-Pro-Max',
    'https://www.whatmobile.com.pk/Apple_iPhone-14-Pro',
    'https://www.whatmobile.com.pk/Apple_iPhone-14-Plus',
    'https://www.whatmobile.com.pk/Apple_iPhone-14'
]

categories = ['Build', 'Frequency', 'Processor', 'Display', 'Memory', 'Camera', 'Connectivity', 'Features', 'Battery', 'Ratings']

mobile_informations = {}

for url in urls:
    name = url[30:].replace('-', ' ').replace('_', ' ')
    html = requests.get(url).text
    soup = BeautifulSoup(html)
    mobile_information = {}
    for row in soup.find('tbody').find_all('tr'):
        for heading, cell in zip(row.find('th'), row.find('td')):
            if cell.text in categories:
                mobile_information[heading.text] = unicodedata.normalize('NFKD', row.find_all('td')[1].text).strip()
                continue
            mobile_information[heading.text] = unicodedata.normalize('NFKD', cell.text).strip()
    for row in soup.find_all('tbody')[1].find_all('tr'):
        mobile_information[row.find('th').find('h2').text.strip()] = unicodedata.normalize('NFKD', row.find('td').find('strong').text).strip()
        break
    mobile_informations[name] = mobile_information
```

```python
with open('mobile_data.json', 'w') as fp:
    json.dump(mobile_informations, fp)
```

```python
with open('mobile_data.json', 'r') as fp:
    mobile_informations = json.load(fp)
```
