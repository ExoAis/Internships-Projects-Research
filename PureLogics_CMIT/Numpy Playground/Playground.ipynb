```python
zibai = np.arange(-361, 361, 5)
```

```python
zibai_math = np.cos(np.radians(zibai))
plt.plot(zibai, zibai_math)
plt.show()
```

```python
zibai_math_sin = np.sin(np.radians(zibai))
plt.plot(zibai, zibai_math_sin)
plt.show()
```

```python
zibai_alter = np.arange(-361, 361, 10)
zibai_math_tan = np.tan(np.radians(zibai_alter))
plt.plot(zibai_alter, zibai_math_tan)
plt.show()
```
