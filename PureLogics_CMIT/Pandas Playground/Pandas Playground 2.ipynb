```python
df_sn = pd.read_csv('../Data Sets/PandasDataset/MulVarLinReg_SomeNull.csv')
df_sn = df_sn.iloc[0 : 5]
df_sn.interpolate('polynomial', order = 2, inplace = True)
df_sn['WindBe'] = df_sn['WindBe'].astype(int)
df_sn.bfill(inplace = True)
df_sn.head()
```

```python
df_nn = pd.read_csv('../Data Sets/PandasDataset/MulVarLinReg_NotAnyNull.csv')
df_nn = df_nn.iloc[0 : 5]
df_nn.head()
```

```python
pd.merge(df_sn, df_nn, on = 'WindBe')
```

```python
pd.merge(df_sn, df_nn, on = 'WindBe', how = 'right')
```

```python
pd.merge(df_sn, df_nn, on = 'WindBe', how = 'right', indicator = True)
```

```python
pd.merge(df_sn, df_nn, on = 'WindBe', how = 'right', indicator = True, suffixes = ['_First', '_Second'])
```

```python
pd.concat([df_sn, df_nn])
```

```python
pd.concat([df_sn, df_nn], axis = 1)
```

```python
pd.concat([df_sn, df_nn], join = 'inner')
```

```python
pd.concat([df_sn, df_nn], ignore_index = True)
```

```python
pd.concat([df_sn, df_nn], keys = ['First', 'Second'], axis = 1)
```

```python
df_sn.join(df_nn, lsuffix = '_First', rsuffix = '_Second')
```

```python
df_sn.join(df_nn, lsuffix = '_First', rsuffix = '_Second', how = 'right')
```
